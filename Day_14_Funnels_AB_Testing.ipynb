```python
import pandas as pd

funnel_data = {
    'Stage': ['Visitors', 'Product Views', 'Add to Cart', 'Purchases'],
    'Users': [1000, 700, 350, 140]
}

funnel_df = pd.DataFrame(funnel_data)

print(funnel_df)
```

```python
funnel_df['Conversion_Rate (%)'] = (
    funnel_df['Users'] / funnel_df['Users'].iloc[0]
) * 100

print(funnel_df)
```

```python
funnel_df['Drop_Off (%)'] = (
    (funnel_df['Users'].shift(1) - funnel_df['Users'])
    / funnel_df['Users'].shift(1)
) * 100

funnel_df.loc[0, 'Drop_Off (%)'] = 0

print(funnel_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(funnel_df['Stage'], funnel_df['Users'])

plt.title('E-commerce Conversion Funnel')
plt.xlabel('Funnel Stage')
plt.ylabel('Number of Users')
plt.xticks(rotation=15)
plt.show()
```

```python
import pandas as pd

ab_data = {
    'Group': ['A', 'B'],
    'Visitors': [1000, 1000],
    'Conversions': [120, 150]
}

ab_df = pd.DataFrame(ab_data)

ab_df['Conversion_Rate (%)'] = (
    ab_df['Conversions'] / ab_df['Visitors']
) * 100

print(ab_df)
```

```python
from scipy.stats import chi2_contingency

observed = [
    [120, 880],
    [150, 850]
]

chi2, p_value, dof, expected = chi2_contingency(observed)

print("Chi-Square Value:", chi2)
print("P-Value:", p_value)
```

```python
if p_value < 0.05:
    print("The result is statistically significant.")
else:
    print("The result is not statistically significant.")

print("Group A Conversion Rate: 12%")
print("Group B Conversion Rate: 15%")
print("Conclusion: More testing may be needed.")
```
