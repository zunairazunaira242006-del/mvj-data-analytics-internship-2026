```python
import pandas as pd
import duckdb

sales = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Tablet', 'Mobile'],
    'Sales': [50000, 20000, 45000, 15000, 25000]
})

result = duckdb.query("""
    SELECT Product, SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Product
    ORDER BY Total_Sales DESC
""").df()

print(result)
```

```python
result = duckdb.query("""
    SELECT
        COUNT(*) AS Total_Records,
        SUM(Sales) AS Overall_Sales,
        AVG(Sales) AS Average_Sales
    FROM sales
""").df()

print(result)
```
