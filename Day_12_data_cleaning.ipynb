```python
import pandas as pd

sales_df = pd.read_csv('/content/online_retail_II.csv')

sales_df.head()
```

```python
sales_df.columns
```

```python
sales_df['TotalPrice'] = sales_df['Quantity'] * sales_df['Price']

sales_df[['Quantity', 'Price', 'TotalPrice']].head()
```

```python
sales_df['InvoiceDate'] = pd.to_datetime(sales_df['InvoiceDate'])

daily_revenue = sales_df.groupby(
    sales_df['InvoiceDate'].dt.date
)['TotalPrice'].sum()

daily_revenue.head()
```

```python
monthly_revenue = sales_df.resample(
    'ME', on='InvoiceDate'
)['TotalPrice'].sum()

monthly_revenue.head()
```

```python
monthly_rolling_avg = monthly_revenue.rolling(window=3).mean()

monthly_rolling_avg.head(10)
```

```python
customer_revenue = sales_df.groupby(
    'Customer ID',
    as_index=False
)['TotalPrice'].sum()

customer_revenue.head()
```

```python
customer_info = sales_df[['Customer ID', 'Country']].drop_duplicates(
    subset='Customer ID'
)

customer_info.head()
```

```python
merged_data = pd.merge(
    customer_revenue,
    customer_info,
    on='Customer ID',
    how='inner'
)

merged_data.head()
```

```python
import pandas as pd

sales_df = pd.read_csv('/content/online_retail_II.csv')

clean_df = sales_df[
    ~sales_df['Invoice'].astype(str).str.startswith('C')
].copy()

print("Original rows:", len(sales_df))
print("After removing cancellations:", len(clean_df))
```

```python
clean_df = clean_df[clean_df['Quantity'] > 0].copy()

print("Rows after removing returns:", len(clean_df))
```

```python
print(clean_df.isnull().sum())
```

```python
clean_df = clean_df.dropna(subset=['Description'])

print("Rows after removing missing descriptions:", len(clean_df))
```

```python
customer_df = clean_df.dropna(subset=['Customer ID']).copy()

print("Rows with Customer ID:", len(customer_df))
```

```python
print(clean_df[['Quantity', 'Price']].describe())
```

```python
print("Negative prices:", (clean_df['Price'] < 0).sum())
print("Quantity above 1000:", (clean_df['Quantity'] > 1000).sum())
print("Price above 1000:", (clean_df['Price'] > 1000).sum())
```

```python
clean_df = clean_df[
    (clean_df['Price'] >= 0) &
    (clean_df['Quantity'] <= 1000) &
    (clean_df['Price'] <= 1000)
].copy()

print("Rows after handling outliers:", len(clean_df))
```

```python
cleaning_log = pd.DataFrame({
    'Cleaning Step': [
        'Cancelled invoices',
        'Product returns',
        'Missing descriptions',
        'Outlier removal'
    ],
    'Records Removed': [
        19494,
        3457,
        1693,
        438
    ]
})

print(cleaning_log)
```

```python
checks = {
    'Missing Invoice': clean_df['Invoice'].isnull().sum(),
    'Missing Description': clean_df['Description'].isnull().sum(),
    'Negative Quantity': (clean_df['Quantity'] < 0).sum(),
    'Negative Price': (clean_df['Price'] < 0).sum(),
    'Duplicate Rows': clean_df.duplicated().sum()
}

for check, result in checks.items():
    print(check, ":", result)
```

```python
clean_df = clean_df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(clean_df))
print("Remaining duplicates:", clean_df.duplicated().sum())
```

```python
cleaning_log.loc[len(cleaning_log)] = ['Duplicate rows', 33820]

checks = {
    'Missing Invoice': clean_df['Invoice'].isnull().sum(),
    'Missing Description': clean_df['Description'].isnull().sum(),
    'Negative Quantity': (clean_df['Quantity'] < 0).sum(),
    'Negative Price': (clean_df['Price'] < 0).sum(),
    'Duplicate Rows': clean_df.duplicated().sum()
}

print(cleaning_log)
print("\nFinal Data Quality Checks:")
for check, result in checks.items():
    print(check, ":", result)
```

```python
clean_df.to_csv('cleaned_online_retail.csv', index=False)
cleaning_log.to_csv('cleaning_log.csv', index=False)

print("Cleaned dataset and cleaning log saved successfully!")
```
